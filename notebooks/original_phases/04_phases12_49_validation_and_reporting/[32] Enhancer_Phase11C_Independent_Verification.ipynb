## Independent verification and statistical interpretation
Every saved checkpoint must reproduce its matching OOF probabilities on identical record IDs, labels and folds. Model identity, pretrained revision/hash, training/evaluation record hashes and protocol/source locks are checked. Every metric is independently recomputed, including confusion counts, AUROC, AUPRC, MCC, accuracy, Brier, sensitivity and specificity.

Seed-pooled performance, mean-fold scores, seed SD, fold SD and three-seed ensemble performance are separate quantities. Fifteen seed/fold combinations are not15independent training samples: fold training sets overlap. Conditional paired intervals resample frozen similarity groups for the fixed ensemble probabilities; they do not estimate the full training algorithm's uncertainty. The six-record mixed-label evaluation exclusion is a separate sensitivity without retraining. Nearconstant predictions are flagged and retained. Historical comparisons remain descriptive and separate from the new primary study.
