## Statistical quantities are not interchangeable
Individual fitted-seed scores, their mean/SD and the three-seed ensemble score are reported separately. The original Phase12 intervals and independently repeated Phase13 intervals retain their own labels and bounds. Internal paired differences compare models on identical records within a protocol; 9B and 9C have different test populations, so there is no paired split-effect estimate.

New filtering comparisons use 1,000 shared bootstrap draws of 1 kb genomic groups unioned by exact/RC identity. These intervals condition on the trained models and examined cohorts. Removed-versus-retained differences are descriptive population contrasts, not an estimate of retraining without filtering. Secondary/subgroup intervals are exploratory unless the original source explicitly supplies multiplicity correction. Constant-vector correlations are undefined and remain explicit.
