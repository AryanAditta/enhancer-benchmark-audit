## Authenticate pretrained models and validate the local runtime
The pinned original pretrained weights are distinct from previously fine-tuned regression/classification weights. Forward/backward checks use two synthetic sequences, take no optimizer steps, and discard all probe weights. Actual jobs use fixed two epochs, training-only RC augmentation, fp16, AdamW2e-5, batch8/accumulation2, no OOF-based selection, and the original seed formula. Full model weights, progress and optimizer/RNG resumption are saved locally.

All60jobs are rerun consistently because original-seed implementation comparability cannot be established. Forty correspond to the declared additional seeds; twenty rerun the original seed. Historical predictions are preserved separately.
