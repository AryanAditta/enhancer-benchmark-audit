```python
import torch
import torch.nn as nn
import numpy as np
import time
import matplotlib.pyplot as plt
from transformers import LongformerTokenizer, LongformerModel
from torch.utils.data import Dataset, DataLoader
from torch.optim import AdamW
from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import accuracy_score, f1_score, confusion_matrix, roc_auc_score, matthews_corrcoef, recall_score, roc_curve, auc
from collections import Counter
import random, re
from torch.cuda.amp import autocast, GradScaler

# --- Reproducibility ---
def set_seed(seed=42):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)
    torch.backends.cudnn.deterministic = True

set_seed()

# --- Utilities ---
def clean_sequence(seq):
    return re.sub('[^ATCG]', '', seq.upper())

def load_fasta_clean(path):
    with open(path) as f:
        lines = f.readlines()
    return [clean_sequence(lines[i+1].strip()) for i in range(0, len(lines), 2)]

def reverse_complement(seq):
    return seq.translate(str.maketrans("ATCG", "TAGC"))[::-1]

def kmer_tokenize(seq, k=6):
    return " ".join([seq[i:i+k] for i in range(len(seq)-k+1)])

def augment_sequence(seq):
    bases = ['A','T','C','G']
    seq_list = list(seq)
    for i in range(len(seq)):
        if random.random() < 0.05:
            seq_list[i] = random.choice([b for b in bases if b != seq_list[i]])
    kmers = ["".join(seq_list[i:i+6]) for i in range(len(seq_list)-5)]
    random.shuffle(kmers)
    return " ".join(kmers)

# --- Dataset ---
class SimpleDataset(Dataset):
    def __init__(self, sequences, labels, tokenizer, max_len=512, augment=False):
        self.sequences = sequences
        self.labels = labels
        self.tokenizer = tokenizer
        self.max_len = max_len
        self.augment = augment

    def __len__(self):
        return len(self.sequences)

    def __getitem__(self, idx):
        seq = self.sequences[idx]
        label = self.labels[idx]
        if self.augment and random.random() < 0.5:
            seq = augment_sequence(seq)
        else:
            seq = kmer_tokenize(seq)
        encoding = self.tokenizer(seq, padding="max_length", truncation=True, max_length=self.max_len, return_tensors="pt")
        return {
            'input_ids': encoding['input_ids'].squeeze(0),
            'attention_mask': encoding['attention_mask'].squeeze(0),
            'label': torch.tensor(label, dtype=torch.float)
        }

# --- Model: Longformer only ---
class LongformerOnly(nn.Module):
    def __init__(self, longformer_name="allenai/longformer-base-4096", dropout=0.3):
        super().__init__()
        self.longformer = LongformerModel.from_pretrained(longformer_name)
        self.classifier = nn.Sequential(
            nn.Dropout(dropout),
            nn.Linear(self.longformer.config.hidden_size, 1)
        )
    def forward(self, input_ids, attention_mask):
        outputs = self.longformer(input_ids=input_ids, attention_mask=attention_mask)
        cls_token = outputs.pooler_output  # [batch_size, hidden_size]
        logits = self.classifier(cls_token)
        return logits.squeeze(1)

# --- Metrics ---
def compute_metrics(y_true, y_pred, y_prob):
    cm = confusion_matrix(y_true, y_pred)
    tn, fp, fn, tp = cm.ravel()
    return {
        'SN': recall_score(y_true, y_pred),
        'SP': tn / (tn + fp + 1e-8),
        'ACC': accuracy_score(y_true, y_pred),
        'F1': f1_score(y_true, y_pred),
        'MCC': matthews_corrcoef(y_true, y_pred),
        'AUC': roc_auc_score(y_true, y_prob)
    }

# --- Training Loop with loss & ROC plotting ---
def train_kfold_with_plots(X, y, stage, tokenizer, epochs=8, k=5):
    skf = StratifiedKFold(n_splits=k, shuffle=True, random_state=42)
    device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
    print(f"[{stage}] Using {torch.cuda.device_count()} GPU(s):", [torch.cuda.get_device_name(i) for i in range(torch.cuda.device_count())])
    scaler = GradScaler()

    all_fprs = []
    all_tprs = []
    all_aucs = []
    mean_fpr = np.linspace(0, 1, 100)

    for fold, (train_idx, val_idx) in enumerate(skf.split(X, y)):
        print(f"\n🔁 {stage} - Fold {fold+1}/{k}")
        train_ds = SimpleDataset([X[i] for i in train_idx], [y[i] for i in train_idx], tokenizer, augment=True)
        val_ds = SimpleDataset([X[i] for i in val_idx], [y[i] for i in val_idx], tokenizer)
        train_loader = DataLoader(train_ds, batch_size=32, shuffle=True)
        val_loader = DataLoader(val_ds, batch_size=64)

        model = LongformerOnly()
        model = nn.DataParallel(model)
        model = model.to(device)

        optimizer = AdamW(model.parameters(), lr=1e-5)
        loss_fn = nn.BCEWithLogitsLoss()

        train_losses = []
        model.train()
        for epoch in range(epochs):
            epoch_loss = 0
            for batch in train_loader:
                optimizer.zero_grad()
                ids = batch['input_ids'].to(device)
                mask = batch['attention_mask'].to(device)
                label = batch['label'].to(device)
                with autocast():
                    logits = model(ids, mask)
                    loss = loss_fn(logits, label)
                scaler.scale(loss).backward()
                scaler.step(optimizer)
                scaler.update()
                epoch_loss += loss.item()
            avg_loss = epoch_loss / len(train_loader)
            train_losses.append(avg_loss)
            print(f"Epoch {epoch+1}/{epochs} Loss: {avg_loss:.4f}")

        # Plot Loss Curve for this fold
        plt.figure(figsize=(6,4))
        plt.plot(range(1, epochs+1), train_losses, marker='o')
        plt.title(f'{stage} Fold {fold+1} Training Loss')
        plt.xlabel('Epoch')
        plt.ylabel('Loss')
        plt.grid()
        plt.show()

        # Evaluation
        model.eval()
        y_true, y_prob, y_pred = [], [], []
        with torch.no_grad():
            for batch in val_loader:
                ids = batch['input_ids'].to(device)
                mask = batch['attention_mask'].to(device)
                label = batch['label'].to(device)
                with autocast():
                    probs = torch.sigmoid(model(ids, mask))
                y_true.extend(label.cpu().numpy())
                y_prob.extend(probs.cpu().numpy())
                y_pred.extend((probs > 0.5).int().cpu().numpy())

        metrics = compute_metrics(y_true, y_pred, y_prob)
        print(f"Fold {fold+1} Metrics:", metrics)

        # ROC curve for this fold
        fpr, tpr, _ = roc_curve(y_true, y_prob)
        roc_auc = auc(fpr, tpr)
        all_fprs.append(np.interp(mean_fpr, fpr, tpr))
        all_fprs[-1][0] = 0.0
        all_aucs.append(roc_auc)

        plt.figure(figsize=(6,6))
        plt.plot(fpr, tpr, label=f'ROC fold {fold+1} (AUC = {roc_auc:.3f})')
        plt.plot([0,1], [0,1], linestyle='--', color='gray')
        plt.xlabel('False Positive Rate')
        plt.ylabel('True Positive Rate')
        plt.title(f'{stage} Fold {fold+1} ROC Curve')
        plt.legend()
        plt.grid()
        plt.show()

    # Plot mean ROC curve
    mean_tpr = np.mean(all_fprs, axis=0)
    mean_tpr[-1] = 1.0
    mean_auc = auc(mean_fpr, mean_tpr)
    std_auc = np.std(all_aucs)

    plt.figure(figsize=(7,7))
    plt.plot(mean_fpr, mean_tpr, color='b',
             label=f'Mean ROC (AUC = {mean_auc:.3f} ± {std_auc:.3f})', lw=2)
    plt.plot([0,1], [0,1], linestyle='--', color='gray')
    plt.xlabel('False Positive Rate')
    plt.ylabel('True Positive Rate')
    plt.title(f'{stage} Mean ROC Curve')
    plt.legend(loc='lower right')
    plt.grid()
    plt.show()

    print(f"\n[{stage}] Mean AUC: {mean_auc:.4f} ± {std_auc:.4f}")

# --- Data Preparation ---
def prepare_data(path1, path2):
    a = load_fasta_clean(path1)
    b = load_fasta_clean(path2)
    max_len = max(len(a), len(b))
    a = (a * ((max_len // len(a)) + 1))[:max_len]
    b = (b * ((max_len // len(b)) + 1))[:max_len]
    X, y = [], []
    for seq in a:
        X += [seq, reverse_complement(seq)]
        y += [1, 1]
    for seq in b:
        X += [seq, reverse_complement(seq)]
        y += [0, 0]
    return X, y

# --- Main run ---
if __name__ == "__main__":
    tokenizer = LongformerTokenizer.from_pretrained("allenai/longformer-base-4096")

    print("\n==== [Stage 1] Enhancer vs Non-Enhancer ====")
    X1, y1 = prepare_data("/kaggle/input/enhancer-detection/enhancer.cv.txt", "/kaggle/input/enhancer-detection/non.cv.txt")
    print("[Stage1] Label Distribution:", Counter(y1))
    train_kfold_with_plots(X1, y1, stage="Stage1", tokenizer=tokenizer)

    print("\n==== [Stage 2] Strong vs Weak Enhancer ====")
    X2, y2 = prepare_data("/kaggle/input/enhancer-detection/strong_742.txt", "/kaggle/input/enhancer-detection/weak_742.txt")
    print("[Stage2] Label Distribution:", Counter(y2))
    train_kfold_with_plots(X2, y2, stage="Stage2", tokenizer=tokenizer)
```

```python
import torch
import torch.nn as nn
import numpy as np
import time
import matplotlib.pyplot as plt
from transformers import LongformerTokenizer, LongformerModel
from torch.utils.data import Dataset, DataLoader
from torch.optim import AdamW
from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import accuracy_score, f1_score, confusion_matrix, roc_auc_score, matthews_corrcoef, recall_score, roc_curve, auc
from collections import Counter
import random, re
from torch.amp import autocast  # ✅ Corrected import
from torch.cuda.amp import GradScaler

# --- Reproducibility ---
def set_seed(seed=42):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)
    torch.backends.cudnn.deterministic = True

set_seed()

# --- Utilities ---
def clean_sequence(seq):
    return re.sub('[^ATCG]', '', seq.upper())

def load_fasta_clean(path):
    with open(path) as f:
        lines = f.readlines()
    return [clean_sequence(lines[i+1].strip()) for i in range(0, len(lines), 2)]

def reverse_complement(seq):
    return seq.translate(str.maketrans("ATCG", "TAGC"))[::-1]

def kmer_tokenize(seq, k=6):
    if len(seq) < k:
        seq += "A" * (k - len(seq))
    return " ".join([seq[i:i+k] for i in range(len(seq)-k+1)])

def augment_sequence(seq):
    bases = ['A','T','C','G']
    seq_list = list(seq)
    for i in range(len(seq_list)):
        if random.random() < 0.05:
            seq_list[i] = random.choice([b for b in bases if b != seq_list[i]])
    if len(seq_list) < 6:
        seq_list += ['A'] * (6 - len(seq_list))
    kmers = ["".join(seq_list[i:i+6]) for i in range(len(seq_list)-5)]
    random.shuffle(kmers)
    return " ".join(kmers)

# --- Dataset ---
class SimpleDataset(Dataset):
    def __init__(self, sequences, labels, tokenizer, max_len=512, augment=False):
        self.sequences = sequences
        self.labels = labels
        self.tokenizer = tokenizer
        self.max_len = max_len
        self.augment = augment

    def __len__(self):
        return len(self.sequences)

    def __getitem__(self, idx):
        seq = self.sequences[idx]
        label = self.labels[idx]
        if self.augment and random.random() < 0.5:
            seq = augment_sequence(seq)
        else:
            seq = kmer_tokenize(seq)
        encoding = self.tokenizer(seq, padding="max_length", truncation=True, max_length=self.max_len, return_tensors="pt")
        return {
            'input_ids': encoding['input_ids'].squeeze(0),
            'attention_mask': encoding['attention_mask'].squeeze(0),
            'label': torch.tensor(label, dtype=torch.float)
        }

# --- Model ---
class Longformer_CNN(nn.Module):
    def __init__(self, longformer_name="allenai/longformer-base-4096", hidden_dim=128, dropout=0.3):
        super().__init__()
        self.longformer = LongformerModel.from_pretrained(longformer_name)
        self.cnn = nn.Sequential(
            nn.Conv1d(self.longformer.config.hidden_size, 64, kernel_size=5, padding=2),
            nn.ReLU(),
            nn.MaxPool1d(kernel_size=2)
        )
        self.classifier = nn.Sequential(
            nn.Linear(64, 128),
            nn.ReLU(),
            nn.Dropout(dropout),
            nn.Linear(128, 1)
        )

    def forward(self, input_ids, attention_mask):
        outputs = self.longformer(input_ids=input_ids, attention_mask=attention_mask)
        hidden_states = outputs.last_hidden_state
        cnn_input = hidden_states.permute(0, 2, 1)
        cnn_out = self.cnn(cnn_input)
        cnn_out = torch.max(cnn_out, dim=2)[0]
        logits = self.classifier(cnn_out)
        return logits.squeeze(1)

# --- Metrics ---
def compute_metrics(y_true, y_pred, y_prob):
    cm = confusion_matrix(y_true, y_pred)
    tn, fp, fn, tp = cm.ravel() if cm.shape == (2, 2) else (0, 0, 0, 0)
    return {
        'SN': recall_score(y_true, y_pred),
        'SP': tn / (tn + fp + 1e-8),
        'ACC': accuracy_score(y_true, y_pred),
        'F1': f1_score(y_true, y_pred),
        'MCC': matthews_corrcoef(y_true, y_pred),
        'AUC': roc_auc_score(y_true, y_prob)
    }

# --- Training ---
def train_kfold_with_plots(X, y, stage, tokenizer, epochs=8, k=5):
    skf = StratifiedKFold(n_splits=k, shuffle=True, random_state=42)
    device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
    print(f"[{stage}] Using {torch.cuda.device_count()} GPU(s):", [torch.cuda.get_device_name(i) for i in range(torch.cuda.device_count())])
    scaler = GradScaler()

    all_fprs = []
    all_aucs = []
    mean_fpr = np.linspace(0, 1, 100)

    for fold, (train_idx, val_idx) in enumerate(skf.split(X, y)):
        print(f"\n🔁 {stage} - Fold {fold+1}/{k}")
        print(f"Val label distribution: {Counter([y[i] for i in val_idx])}")
        train_ds = SimpleDataset([X[i] for i in train_idx], [y[i] for i in train_idx], tokenizer, augment=True)
        val_ds = SimpleDataset([X[i] for i in val_idx], [y[i] for i in val_idx], tokenizer)
        train_loader = DataLoader(train_ds, batch_size=16, shuffle=True)
        val_loader = DataLoader(val_ds, batch_size=32)

        model = Longformer_CNN()
        model = nn.DataParallel(model)
        model = model.to(device)

        optimizer = AdamW(model.parameters(), lr=5e-6)
        loss_fn = nn.BCEWithLogitsLoss()

        train_losses = []
        model.train()
        for epoch in range(epochs):
            epoch_loss = 0
            for batch in train_loader:
                optimizer.zero_grad()
                ids = batch['input_ids'].to(device)
                mask = batch['attention_mask'].to(device)
                label = batch['label'].to(device)

                with autocast(device_type='cuda'):
                    logits = model(ids, mask)
                    if torch.isnan(logits).any() or torch.isinf(logits).any():
                        print("⚠️ Invalid logits detected.")
                        continue
                    loss = loss_fn(logits, label)

                if torch.isnan(loss) or torch.isinf(loss):
                    print("NaN or Inf loss. Skipping this batch.")
                    continue

                scaler.scale(loss).backward()
                torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)
                scaler.step(optimizer)
                scaler.update()
                epoch_loss += loss.item()

            avg_loss = epoch_loss / len(train_loader)
            train_losses.append(avg_loss)
            print(f"Epoch {epoch+1}/{epochs} Loss: {avg_loss:.4f}")

        # Plot training loss
        plt.figure()
        plt.plot(range(1, epochs+1), train_losses, marker='o')
        plt.title(f'{stage} Fold {fold+1} Training Loss')
        plt.xlabel('Epoch')
        plt.ylabel('Loss')
        plt.grid()
        plt.show()

        # Validation
        model.eval()
        y_true, y_prob, y_pred = [], [], []
        with torch.no_grad():
            for batch in val_loader:
                ids = batch['input_ids'].to(device)
                mask = batch['attention_mask'].to(device)
                label = batch['label'].to(device)
                with autocast(device_type='cuda'):
                    probs = torch.sigmoid(model(ids, mask))
                y_true.extend(label.cpu().numpy())
                y_prob.extend(probs.cpu().numpy())
                y_pred.extend((probs > 0.5).int().cpu().numpy())

        # Skip if only one class
        if len(set(y_true)) < 2:
            print("⚠️ Only one class present. Skipping metrics and ROC.")
            continue

        print(f"Fold {fold+1} Metrics:", compute_metrics(y_true, y_pred, y_prob))

        fpr, tpr, _ = roc_curve(y_true, y_prob)
        roc_auc = auc(fpr, tpr)
        all_fprs.append(np.interp(mean_fpr, fpr, tpr))
        all_aucs.append(roc_auc)

        plt.figure()
        plt.plot(fpr, tpr, label=f'ROC fold {fold+1} (AUC = {roc_auc:.3f})')
        plt.plot([0,1], [0,1], linestyle='--', color='gray')
        plt.xlabel('False Positive Rate')
        plt.ylabel('True Positive Rate')
        plt.title(f'{stage} Fold {fold+1} ROC Curve')
        plt.legend()
        plt.grid()
        plt.show()

    # Mean ROC
    if all_fprs:
        mean_tpr = np.mean(all_fprs, axis=0)
        mean_tpr[-1] = 1.0
        mean_auc = auc(mean_fpr, mean_tpr)
        std_auc = np.std(all_aucs)

        plt.figure(figsize=(7,7))
        plt.plot(mean_fpr, mean_tpr, color='b', label=f'Mean ROC (AUC = {mean_auc:.3f} ± {std_auc:.3f})', lw=2)
        plt.plot([0,1], [0,1], linestyle='--', color='gray')
        plt.xlabel('False Positive Rate')
        plt.ylabel('True Positive Rate')
        plt.title(f'{stage} Mean ROC Curve')
        plt.legend(loc='lower right')
        plt.grid()
        plt.show()

        print(f"\n[{stage}] Mean AUC: {mean_auc:.4f} ± {std_auc:.4f}")

# --- Data Preparation ---
def prepare_data(path1, path2):
    a = load_fasta_clean(path1)
    b = load_fasta_clean(path2)
    X, y = [], []
    for seq in a:
        X += [seq, reverse_complement(seq)]
        y += [1, 1]
    for seq in b:
        X += [seq, reverse_complement(seq)]
        y += [0, 0]
    return X, y

# --- Main Run ---
if __name__ == "__main__":
    tokenizer = LongformerTokenizer.from_pretrained("allenai/longformer-base-4096")

    print("\n==== [Stage 2] Strong vs Weak Enhancer ====")
    X2, y2 = prepare_data("/kaggle/input/enhancer-detection/strong_742.txt", "/kaggle/input/enhancer-detection/weak_742.txt")
    print("[Stage2] Label Distribution:", Counter(y2))
    train_kfold_with_plots(X2, y2, stage="Stage2", tokenizer=tokenizer)

    print("\n==== [Stage 1] Enhancer vs Non-Enhancer ====")
    X1, y1 = prepare_data("/kaggle/input/enhancer-detection/enhancer.cv.txt", "/kaggle/input/enhancer-detection/non.cv.txt")
    print("[Stage1] Label Distribution:", Counter(y1))
    train_kfold_with_plots(X1, y1, stage="Stage1", tokenizer=tokenizer)
```

```python
import torch
import torch.nn as nn
import numpy as np
import time
import matplotlib.pyplot as plt
from transformers import LongformerTokenizer, LongformerModel
from torch.utils.data import Dataset, DataLoader
from torch.optim import AdamW
from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import accuracy_score, f1_score, confusion_matrix, roc_auc_score, matthews_corrcoef, recall_score, roc_curve, auc
from collections import Counter
import random, re
from torch.cuda.amp import autocast, GradScaler
import seaborn as sns
import os

# Enable expandable segments to mitigate memory fragmentation
os.environ["PYTORCH_CUDA_ALLOC_CONF"] = "expandable_segments:True"

# --- Reproducibility ---
def set_seed(seed=42):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)
    torch.backends.cudnn.deterministic = True

set_seed()

# --- Utilities ---
def clean_sequence(seq):
    return re.sub('[^ATCG]', '', seq.upper())

def load_fasta_clean(path):
    with open(path) as f:
        lines = f.readlines()
    return [clean_sequence(lines[i+1].strip()) for i in range(0, len(lines), 2)]

def reverse_complement(seq):
    return seq.translate(str.maketrans("ATCG", "TAGC"))[::-1]

def kmer_tokenize(seq, k=6):
    return " ".join([seq[i:i+k] for i in range(len(seq)-k+1)])

def augment_sequence(seq):
    bases = ['A','T','C','G']
    seq_list = list(seq)
    for i in range(len(seq)):
        if random.random() < 0.05:
            seq_list[i] = random.choice([b for b in bases if b != seq_list[i]])
    kmers = ["".join(seq_list[i:i+6]) for i in range(len(seq_list)-5)]
    random.shuffle(kmers)
    return " ".join(kmers)

# --- Dataset ---
class SimpleDataset(Dataset):
    def __init__(self, sequences, labels, tokenizer, max_len=128, augment=False):
        self.sequences = sequences
        self.labels = labels
        self.tokenizer = tokenizer
        self.max_len = max_len
        self.augment = augment

    def __len__(self):
        return len(self.sequences)

    def __getitem__(self, idx):
        seq = self.sequences[idx]
        label = self.labels[idx]
        if self.augment and random.random() < 0.5:
            seq = augment_sequence(seq)
        else:
            seq = kmer_tokenize(seq)
        encoding = self.tokenizer(seq, padding="max_length", truncation=True, max_length=self.max_len, return_tensors="pt")
        return {
            'input_ids': encoding['input_ids'].squeeze(0),
            'attention_mask': encoding['attention_mask'].squeeze(0),
            'label': torch.tensor(label, dtype=torch.float)
        }

# --- Longformer + BiLSTM Model ---
class Longformer_BiLSTM(nn.Module):
    def __init__(self, longformer_name="allenai/longformer-base-4096", hidden_dim=128, num_layers=2, dropout=0.3):
        super().__init__()
        self.longformer = LongformerModel.from_pretrained(longformer_name)
        self.lstm = nn.LSTM(self.longformer.config.hidden_size, hidden_dim, num_layers=num_layers, batch_first=True, bidirectional=True)
        self.dropout = nn.Dropout(dropout)
        self.classifier = nn.Linear(hidden_dim * 2, 1)

    def forward(self, input_ids, attention_mask):
        outputs = self.longformer(input_ids=input_ids, attention_mask=attention_mask)
        hidden_states = outputs.last_hidden_state
        lstm_out, _ = self.lstm(hidden_states)
        pooled = torch.mean(lstm_out, dim=1)
        logits = self.classifier(self.dropout(pooled)).squeeze(-1)
        return logits

# --- Evaluation & Plotting ---
def compute_metrics(y_true, y_pred, y_prob):
    cm = confusion_matrix(y_true, y_pred)
    tn, fp, fn, tp = cm.ravel()
    return {
        'SN': recall_score(y_true, y_pred),
        'SP': tn / (tn + fp),
        'ACC': accuracy_score(y_true, y_pred),
        'F1': f1_score(y_true, y_pred),
        'MCC': matthews_corrcoef(y_true, y_pred),
        'AUC': roc_auc_score(y_true, y_prob),
        'CM': cm
    }

def plot_auc_curve(y_true, y_prob, stage, fold):
    fpr, tpr, _ = roc_curve(y_true, y_prob)
    plt.figure()
    plt.plot(fpr, tpr, label=f'ROC curve (AUC = {roc_auc_score(y_true, y_prob):.4f})')
    plt.plot([0, 1], [0, 1], 'k--')
    plt.xlabel('False Positive Rate')
    plt.ylabel('True Positive Rate')
    plt.title(f'{stage} - Fold {fold+1} ROC Curve')
    plt.legend(loc='lower right')
    plt.savefig(f'{stage}_fold{fold+1}_roc.png')
    plt.close()

def plot_loss_curve(losses, stage, fold):
    plt.figure()
    plt.plot(losses, label='Training Loss')
    plt.title(f'{stage} - Fold {fold+1} Loss Curve')
    plt.xlabel('Epoch')
    plt.ylabel('Loss')
    plt.legend()
    plt.savefig(f'{stage}_fold{fold+1}_loss.png')
    plt.close()

def plot_confusion_matrix(cm, stage, fold):
    plt.figure()
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=['Pred 0', 'Pred 1'], yticklabels=['True 0', 'True 1'])
    plt.title(f'{stage} - Fold {fold+1} Confusion Matrix')
    plt.savefig(f'{stage}_fold{fold+1}_cm.png')
    plt.close()

# --- Data Prep ---
def prepare_stage_data(stage):
    if stage == "Stage1":
        enhancers = load_fasta_clean("/kaggle/input/enhancer-detection/enhancer.cv.txt")
        non_enhancers = load_fasta_clean("/kaggle/input/enhancer-detection/non.cv.txt")
        enhancers = enhancers[:len(non_enhancers)]
        X = enhancers + [reverse_complement(seq) for seq in enhancers] + \
            non_enhancers + [reverse_complement(seq) for seq in non_enhancers]
        y = [1]*len(enhancers)*2 + [0]*len(non_enhancers)*2
    else:
        strong = load_fasta_clean("/kaggle/input/enhancer-detection/strong_742.txt")
        weak = load_fasta_clean("/kaggle/input/enhancer-detection/weak_742.txt")
        max_len = max(len(strong), len(weak))
        strong = (strong * ((max_len // len(strong)) + 1))[:max_len]
        weak = (weak * ((max_len // len(weak)) + 1))[:max_len]
        X = strong + [reverse_complement(seq) for seq in strong] + \
            weak + [reverse_complement(seq) for seq in weak]
        y = [1]*len(strong)*2 + [0]*len(weak)*2
    return X, y

# --- Training ---
def run_cross_validation(X, y, stage, tokenizer, epochs=5, k=5):
    skf = StratifiedKFold(n_splits=k, shuffle=True, random_state=42)
    device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
    results = []

    for fold, (train_idx, val_idx) in enumerate(skf.split(X, y)):
        print(f"\n🔁 {stage} - Fold {fold+1}/{k}")
        train_dataset = SimpleDataset([X[i] for i in train_idx], [y[i] for i in train_idx], tokenizer, augment=True)
        val_dataset = SimpleDataset([X[i] for i in val_idx], [y[i] for i in val_idx], tokenizer)

        train_loader = DataLoader(train_dataset, batch_size=16, shuffle=True)
        val_loader = DataLoader(val_dataset, batch_size=32)

        model = Longformer_BiLSTM()
        model = nn.DataParallel(model)
        model = model.to(device)

        optimizer = AdamW(model.parameters(), lr=1e-5)
        loss_fn = nn.BCEWithLogitsLoss()
        scaler = GradScaler()

        losses = []
        start_time = time.time()

        for epoch in range(epochs):
            model.train()
            total_loss = 0
            for batch in train_loader:
                optimizer.zero_grad()
                input_ids = batch['input_ids'].to(device)
                attention_mask = batch['attention_mask'].to(device)
                labels = batch['label'].to(device)

                with autocast():
                    logits = model(input_ids, attention_mask)
                    loss = loss_fn(logits, labels)
                scaler.scale(loss).backward()
                scaler.step(optimizer)
                scaler.update()
                total_loss += loss.item()

            avg_loss = total_loss / len(train_loader)
            losses.append(avg_loss)
            print(f"Epoch {epoch+1} Loss: {avg_loss:.4f}")

        train_time = time.time() - start_time
        memory_used = torch.cuda.max_memory_allocated() / 1e6
        param_count = sum(p.numel() for p in model.parameters() if p.requires_grad)

        model.eval()
        y_true, y_pred, y_prob = [], [], []
        inference_start = time.time()
        with torch.no_grad():
            for batch in val_loader:
                input_ids = batch['input_ids'].to(device)
                attention_mask = batch['attention_mask'].to(device)
                labels = batch['label'].to(device)
                with autocast():
                    logits = model(input_ids, attention_mask)
                    probs = torch.sigmoid(logits)
                y_true.extend(labels.cpu().tolist())
                y_prob.extend(probs.cpu().tolist())
                y_pred.extend((probs > 0.5).int().cpu().tolist())
        inference_time = time.time() - inference_start

        metrics = compute_metrics(y_true, y_pred, y_prob)
        metrics.update({
            'TrainTimeSec': train_time,
            'InferTimeSec': inference_time,
            'Params': param_count,
            'MemMB': memory_used
        })
        print(metrics)

        plot_auc_curve(y_true, y_prob, stage, fold)
        plot_loss_curve(losses, stage, fold)
        plot_confusion_matrix(metrics['CM'], stage, fold)
        results.append(metrics)

    avg = {k: np.mean([fold[k] for fold in results if k in fold]) for k in results[0]}
    print(f"\n=== Final {stage} {k}-Fold Averages ===")
    for k, v in avg.items():
        print(f"{k}: {v:.4f}")

# --- Run ---
if __name__ == "__main__":
    tokenizer = LongformerTokenizer.from_pretrained("allenai/longformer-base-4096")
    for stage in ["Stage1", "Stage2"]:
        X, y = prepare_stage_data(stage)
        run_cross_validation(X, y, stage, tokenizer, epochs=5, k=5)
```

```python
import torch
import torch.nn as nn
import numpy as np
import time
import matplotlib.pyplot as plt
from transformers import LongformerTokenizer, LongformerModel
from torch.utils.data import Dataset, DataLoader
from torch.optim import AdamW
from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import accuracy_score, f1_score, confusion_matrix, roc_auc_score, matthews_corrcoef, recall_score, roc_curve, auc
from collections import Counter
import random, re
from torch.cuda.amp import autocast, GradScaler
import os

os.environ["PYTORCH_CUDA_ALLOC_CONF"] = "expandable_segments:True"

# --- Reproducibility ---
def set_seed(seed=42):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)
    torch.backends.cudnn.deterministic = True

set_seed()

# --- Utilities ---
def clean_sequence(seq):
    return re.sub('[^ATCG]', '', seq.upper())

def load_fasta_clean(path):
    with open(path) as f:
        lines = f.readlines()
    return [clean_sequence(lines[i+1].strip()) for i in range(0, len(lines), 2)]

def reverse_complement(seq):
    return seq.translate(str.maketrans("ATCG", "TAGC"))[::-1]

def kmer_tokenize(seq, k=6):
    return " ".join([seq[i:i+k] for i in range(len(seq)-k+1)])

def augment_sequence(seq):
    bases = ['A','T','C','G']
    seq_list = list(seq)
    for i in range(len(seq)):
        if random.random() < 0.05:
            seq_list[i] = random.choice([b for b in bases if b != seq_list[i]])
    kmers = ["".join(seq_list[i:i+6]) for i in range(len(seq_list)-5)]
    random.shuffle(kmers)
    return " ".join(kmers)

# --- Dataset ---
class SimpleDataset(Dataset):
    def __init__(self, sequences, labels, tokenizer, max_len=128, augment=False):
        self.sequences = sequences
        self.labels = labels
        self.tokenizer = tokenizer
        self.max_len = max_len
        self.augment = augment

    def __len__(self):
        return len(self.sequences)

    def __getitem__(self, idx):
        seq = self.sequences[idx]
        label = self.labels[idx]
        if self.augment and random.random() < 0.5:
            seq = augment_sequence(seq)
        else:
            seq = kmer_tokenize(seq)
        encoding = self.tokenizer(seq, padding="max_length", truncation=True, max_length=self.max_len, return_tensors="pt")
        return {
            'input_ids': encoding['input_ids'].squeeze(0),
            'attention_mask': encoding['attention_mask'].squeeze(0),
            'label': torch.tensor(label, dtype=torch.float)
        }

# --- Model ---
class Longformer_TransformerEncoder(nn.Module):
    def __init__(self, longformer_name="allenai/longformer-base-4096", hidden_dim=128, num_layers=2, dropout=0.3):
        super().__init__()
        self.longformer = LongformerModel.from_pretrained(longformer_name)
        encoder_layer = nn.TransformerEncoderLayer(
            d_model=self.longformer.config.hidden_size,
            nhead=8,
            dim_feedforward=hidden_dim,
            dropout=dropout
        )
        self.transformer_encoder = nn.TransformerEncoder(encoder_layer, num_layers=num_layers)
        self.classifier = nn.Sequential(
            nn.Linear(self.longformer.config.hidden_size, 128),
            nn.ReLU(),
            nn.Dropout(dropout),
            nn.Linear(128, 1)
        )

    def forward(self, input_ids, attention_mask):
        outputs = self.longformer(input_ids=input_ids, attention_mask=attention_mask)
        hidden_states = outputs.last_hidden_state
        encoded = self.transformer_encoder(hidden_states)
        cls_token = encoded[:, 0, :]
        logits = self.classifier(cls_token)
        return logits.squeeze(1)

# --- Metrics ---
def compute_metrics(y_true, y_pred, y_prob):
    cm = confusion_matrix(y_true, y_pred)
    tn, fp, fn, tp = cm.ravel()
    return {
        'SN': recall_score(y_true, y_pred),
        'SP': tn / (tn + fp + 1e-8),
        'ACC': accuracy_score(y_true, y_pred),
        'F1': f1_score(y_true, y_pred),
        'MCC': matthews_corrcoef(y_true, y_pred),
        'AUC': roc_auc_score(y_true, y_prob)
    }

# --- Training ---
def train_kfold_with_plots(X, y, stage, tokenizer, epochs=8, k=5):
    skf = StratifiedKFold(n_splits=k, shuffle=True, random_state=42)
    device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
    print(f"[{stage}] Using {torch.cuda.device_count()} GPU(s):", [torch.cuda.get_device_name(i) for i in range(torch.cuda.device_count())])
    scaler = GradScaler()

    all_fprs = []
    all_aucs = []
    mean_fpr = np.linspace(0, 1, 100)

    for fold, (train_idx, val_idx) in enumerate(skf.split(X, y)):
        print(f"\n🔁 {stage} - Fold {fold+1}/{k}")
        train_ds = SimpleDataset([X[i] for i in train_idx], [y[i] for i in train_idx], tokenizer, augment=True)
        val_ds = SimpleDataset([X[i] for i in val_idx], [y[i] for i in val_idx], tokenizer)
        train_loader = DataLoader(train_ds, batch_size=8, shuffle=True)
        val_loader = DataLoader(val_ds, batch_size=16)

        model = Longformer_TransformerEncoder()
        model = nn.DataParallel(model)
        model = model.to(device)

        optimizer = AdamW(model.parameters(), lr=1e-5)
        loss_fn = nn.BCEWithLogitsLoss()

        train_losses = []
        model.train()
        for epoch in range(epochs):
            epoch_loss = 0
            for batch in train_loader:
                optimizer.zero_grad()
                ids = batch['input_ids'].to(device)
                mask = batch['attention_mask'].to(device)
                label = batch['label'].to(device)
                with autocast():
                    logits = model(ids, mask)
                    loss = loss_fn(logits, label)
                scaler.scale(loss).backward()
                scaler.step(optimizer)
                scaler.update()
                epoch_loss += loss.item()
            avg_loss = epoch_loss / len(train_loader)
            train_losses.append(avg_loss)
            print(f"Epoch {epoch+1}/{epochs} Loss: {avg_loss:.4f}")

        # Evaluation
        model.eval()
        y_true, y_prob, y_pred = [], [], []
        with torch.no_grad():
            for batch in val_loader:
                ids = batch['input_ids'].to(device)
                mask = batch['attention_mask'].to(device)
                label = batch['label'].to(device)
                with autocast():
                    probs = torch.sigmoid(model(ids, mask))
                if torch.isnan(probs).any():
                    print(f"🚨 NaNs encountered in fold {fold+1}, skipping this fold.")
                    continue
                y_true.extend(label.cpu().numpy())
                y_prob.extend(probs.cpu().numpy())
                y_pred.extend((probs > 0.5).int().cpu().numpy())

        if not y_prob:
            print(f"🚨 No predictions for fold {fold+1}, skipping.")
            continue

        fpr, tpr, _ = roc_curve(y_true, y_prob)
        roc_auc = auc(fpr, tpr)
        interp_tpr = np.interp(mean_fpr, fpr, tpr)
        interp_tpr[-1] = 1.0
        all_fprs.append(interp_tpr)
        all_aucs.append(roc_auc)

        print(f"Fold {fold+1} Metrics:", compute_metrics(y_true, y_pred, y_prob))

        plt.figure(figsize=(6,6))
        plt.plot(fpr, tpr, label=f'ROC fold {fold+1} (AUC = {roc_auc:.3f})')
        plt.plot([0,1], [0,1], linestyle='--', color='gray')
        plt.xlabel('False Positive Rate')
        plt.ylabel('True Positive Rate')
        plt.title(f'{stage} Fold {fold+1} ROC Curve')
        plt.legend()
        plt.grid()
        plt.show()

    # --- Mean ROC ---
    if all_fprs:
        mean_tpr = np.mean(all_fprs, axis=0)
        mean_auc = auc(mean_fpr, mean_tpr)
        std_auc = np.std(all_aucs)

        plt.figure(figsize=(7,7))
        plt.plot(mean_fpr, mean_tpr, color='b', label=f'Mean ROC (AUC = {mean_auc:.3f} ± {std_auc:.3f})', lw=2)
        plt.plot([0,1], [0,1], linestyle='--', color='gray')
        plt.xlabel('False Positive Rate')
        plt.ylabel('True Positive Rate')
        plt.title(f'{stage} Mean ROC Curve')
        plt.legend(loc='lower right')
        plt.grid()
        plt.show()

        print(f"\n[{stage}] Mean AUC: {mean_auc:.4f} ± {std_auc:.4f}")
    else:
        print(f"\n⚠️ [{stage}] No valid ROC curves were collected. Skipping mean ROC curve.")

# --- Data Prep ---
def prepare_data(path1, path2):
    a = load_fasta_clean(path1)
    b = load_fasta_clean(path2)
    max_len = max(len(a), len(b))
    a = (a * ((max_len // len(a)) + 1))[:max_len]
    b = (b * ((max_len // len(b)) + 1))[:max_len]
    X, y = [], []
    for seq in a:
        X += [seq, reverse_complement(seq)]
        y += [1, 1]
    for seq in b:
        X += [seq, reverse_complement(seq)]
        y += [0, 0]
    return X, y

# --- Run ---
if __name__ == "__main__":
    tokenizer = LongformerTokenizer.from_pretrained("allenai/longformer-base-4096")

    print("\n==== [Stage 1] Enhancer vs Non-Enhancer ====")
    X1, y1 = prepare_data("/kaggle/input/enhancer-detection/enhancer.cv.txt", "/kaggle/input/enhancer-detection/non.cv.txt")
    print("[Stage1] Label Distribution:", Counter(y1))
    train_kfold_with_plots(X1, y1, stage="Stage1", tokenizer=tokenizer)

    print("\n==== [Stage 2] Strong vs Weak Enhancer ====")
    X2, y2 = prepare_data("/kaggle/input/enhancer-detection/strong_742.txt", "/kaggle/input/enhancer-detection/weak_742.txt")
    print("[Stage2] Label Distribution:", Counter(y2))
    train_kfold_with_plots(X2, y2, stage="Stage2", tokenizer=tokenizer)
```

```python
import torch
import torch.nn as nn
import numpy as np
import time
import matplotlib.pyplot as plt
from transformers import LongformerTokenizer, LongformerModel
from torch.utils.data import Dataset, DataLoader
from torch.optim import AdamW
from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import accuracy_score, f1_score, confusion_matrix, roc_auc_score, matthews_corrcoef, recall_score, roc_curve, auc
from collections import Counter
import random, re, os, gc
from torch.cuda.amp import autocast, GradScaler

# Enable expandable segments to mitigate memory fragmentation
os.environ["PYTORCH_CUDA_ALLOC_CONF"] = "expandable_segments:True"

# --- Reproducibility ---
def set_seed(seed=42):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)
    torch.backends.cudnn.deterministic = True

set_seed()

# --- Utilities ---
def clean_sequence(seq):
    return re.sub('[^ATCG]', '', seq.upper())

def load_fasta_clean(path):
    with open(path) as f:
        lines = f.readlines()
    return [clean_sequence(lines[i+1].strip()) for i in range(0, len(lines), 2)]

def reverse_complement(seq):
    return seq.translate(str.maketrans("ATCG", "TAGC"))[::-1]

def kmer_tokenize(seq, k=6):
    return " ".join([seq[i:i+k] for i in range(len(seq)-k+1)])

def augment_sequence(seq):
    bases = ['A','T','C','G']
    seq_list = list(seq)
    for i in range(len(seq)):
        if random.random() < 0.05:
            seq_list[i] = random.choice([b for b in bases if b != seq_list[i]])
    kmers = ["".join(seq_list[i:i+6]) for i in range(len(seq_list)-5)]
    random.shuffle(kmers)
    return " ".join(kmers)

# --- Dataset ---
class SimpleDataset(Dataset):
    def __init__(self, sequences, labels, tokenizer, max_len=128, augment=False):
        self.sequences = sequences
        self.labels = labels
        self.tokenizer = tokenizer
        self.max_len = max_len
        self.augment = augment

    def __len__(self):
        return len(self.sequences)

    def __getitem__(self, idx):
        seq = self.sequences[idx]
        label = self.labels[idx]
        if self.augment and random.random() < 0.5:
            seq = augment_sequence(seq)
        else:
            seq = kmer_tokenize(seq)
        encoding = self.tokenizer(seq, padding="max_length", truncation=True, max_length=self.max_len, return_tensors="pt")
        return {
            'input_ids': encoding['input_ids'].squeeze(0),
            'attention_mask': encoding['attention_mask'].squeeze(0),
            'label': torch.tensor(label, dtype=torch.float)
        }

# --- Model: Longformer + CNN + BiLSTM + Attention ---
class Longformer_CNN_BiLSTM_Attention(nn.Module):
    def __init__(self, longformer_name="allenai/longformer-base-4096", hidden_dim=128, dropout=0.3):
        super().__init__()
        self.longformer = LongformerModel.from_pretrained(longformer_name)
        self.cnn = nn.Sequential(
            nn.Conv1d(self.longformer.config.hidden_size, 64, kernel_size=5, padding=2),
            nn.ReLU(),
            nn.MaxPool1d(kernel_size=2)
        )
        self.lstm = nn.LSTM(self.longformer.config.hidden_size, hidden_dim, batch_first=True, bidirectional=True)
        self.attn_fc = nn.Linear(hidden_dim * 2, 1)
        self.classifier = nn.Sequential(
            nn.Linear(self.longformer.config.hidden_size + 64 + hidden_dim * 2, 128),
            nn.ReLU(),
            nn.Dropout(dropout),
            nn.Linear(128, 1)
        )

    def forward(self, input_ids, attention_mask):
        outputs = self.longformer(input_ids=input_ids, attention_mask=attention_mask)
        cls_token = outputs.pooler_output
        hidden_states = outputs.last_hidden_state

        cnn_input = hidden_states.permute(0, 2, 1)
        cnn_out = self.cnn(cnn_input)
        cnn_out = torch.max(cnn_out, dim=2)[0]

        lstm_out, _ = self.lstm(hidden_states)
        attn_scores = torch.softmax(self.attn_fc(lstm_out).squeeze(-1), dim=1)
        attn_out = torch.sum(attn_scores.unsqueeze(-1) * lstm_out, dim=1)

        combined = torch.cat([cls_token, cnn_out, attn_out], dim=1)
        logits = self.classifier(combined)
        return logits.squeeze(1)

# --- Metrics ---
def compute_metrics(y_true, y_pred, y_prob):
    cm = confusion_matrix(y_true, y_pred)
    tn, fp, fn, tp = cm.ravel()
    return {
        'SN': recall_score(y_true, y_pred),
        'SP': tn / (tn + fp + 1e-8),
        'ACC': accuracy_score(y_true, y_pred),
        'F1': f1_score(y_true, y_pred),
        'MCC': matthews_corrcoef(y_true, y_pred),
        'AUC': roc_auc_score(y_true, y_prob)
    }

# --- Training Loop with Loss & ROC plotting ---
def train_kfold_with_plots(X, y, stage, tokenizer, epochs=8, k=5):
    skf = StratifiedKFold(n_splits=k, shuffle=True, random_state=42)
    device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
    print(f"[{stage}] Using {torch.cuda.device_count()} GPU(s):", [torch.cuda.get_device_name(i) for i in range(torch.cuda.device_count())])
    scaler = GradScaler()

    all_fprs = []
    all_tprs = []
    all_aucs = []
    mean_fpr = np.linspace(0, 1, 100)

    batch_size_train = 8  # Reduce batch size for training
    batch_size_val = 16  # Reduce batch size for validation
    max_len = 128  # Reduce sequence length

    for fold, (train_idx, val_idx) in enumerate(skf.split(X, y)):
        print(f"\n🔁 {stage} - Fold {fold+1}/{k}")
        train_ds = SimpleDataset([X[i] for i in train_idx], [y[i] for i in train_idx], tokenizer, max_len=max_len, augment=True)
        val_ds = SimpleDataset([X[i] for i in val_idx], [y[i] for i in val_idx], tokenizer, max_len=max_len)
        train_loader = DataLoader(train_ds, batch_size=batch_size_train, shuffle=True)
        val_loader = DataLoader(val_ds, batch_size=batch_size_val)

        model = Longformer_CNN_BiLSTM_Attention()
        model = nn.DataParallel(model)
        model = model.to(device)

        optimizer = AdamW(model.parameters(), lr=1e-5)
        loss_fn = nn.BCEWithLogitsLoss()

        train_losses = []
        model.train()
        for epoch in range(epochs):
            epoch_loss = 0
            for batch in train_loader:
                optimizer.zero_grad()
                ids = batch['input_ids'].to(device)
                mask = batch['attention_mask'].to(device)
                label = batch['label'].to(device)
                with autocast():
                    logits = model(ids, mask)
                    loss = loss_fn(logits, label)
                scaler.scale(loss).backward()
                scaler.step(optimizer)
                scaler.update()
                epoch_loss += loss.item()
            avg_loss = epoch_loss / len(train_loader)
            train_losses.append(avg_loss)
            print(f"Epoch {epoch+1}/{epochs} Loss: {avg_loss:.4f}")

            # Plot Loss Curve for this fold
            plt.figure(figsize=(6,4))
            plt.plot(range(1, epoch+2), train_losses, marker='o')
            plt.title(f'{stage} Fold {fold+1} Training Loss')
            plt.xlabel('Epoch')
            plt.ylabel('Loss')
            plt.grid()
            plt.show()

        # Evaluation
        model.eval()
        y_true, y_prob, y_pred = [], [], []
        with torch.no_grad():
            infer_start = time.time()
            for batch in val_loader:
                ids = batch['input_ids'].to(device)
                mask = batch['attention_mask'].to(device)
                label = batch['label'].to(device)
                with autocast():
                    probs = torch.sigmoid(model(ids, mask))
                y_true.extend(label.cpu().numpy())
                y_prob.extend(probs.cpu().numpy())
                y_pred.extend((probs > 0.5).int().cpu().numpy())
            infer_time = time.time() - infer_start
        print(f"Inference Time (Fold {fold+1}): {infer_time:.2f} sec")
        print(f"Fold {fold+1} Metrics:", compute_metrics(y_true, y_pred, y_prob))

        # ROC curve for this fold
        fpr, tpr, _ = roc_curve(y_true, y_prob)
        roc_auc = auc(fpr, tpr)
        all_fprs.append(np.interp(mean_fpr, fpr, tpr))
        all_fprs[-1][0] = 0.0
        all_aucs.append(roc_auc)

        plt.figure(figsize=(6,6))
        plt.plot(fpr, tpr, label=f'ROC fold {fold+1} (AUC = {roc_auc:.3f})')
        plt.plot([0,1], [0,1], linestyle='--', color='gray')
        plt.xlabel('False Positive Rate')
        plt.ylabel('True Positive Rate')
        plt.title(f'{stage} Fold {fold+1} ROC Curve')
        plt.legend()
        plt.grid()
        plt.show()

    # Plot mean ROC curve
    mean_tpr = np.mean(all_fprs, axis=0)
    mean_tpr[-1] = 1.0
    mean_auc = auc(mean_fpr, mean_tpr)
    std_auc = np.std(all_aucs)

    plt.figure(figsize=(7,7))
    plt.plot(mean_fpr, mean_tpr, color='b', label=f'Mean ROC (AUC = {mean_auc:.3f} ± {std_auc:.3f})', lw=2)
    plt.plot([0,1], [0,1], linestyle='--', color='gray')
    plt.xlabel('False Positive Rate')
    plt.ylabel('True Positive Rate')
    plt.title(f'{stage} Mean ROC Curve')
    plt.legend(loc='lower right')
    plt.grid()
    plt.show()

    print(f"\n[{stage}] Mean AUC: {mean_auc:.4f} ± {std_auc:.4f}")

# --- Data Preparation ---
def prepare_data(path1, path2):
    a = load_fasta_clean(path1)
    b = load_fasta_clean(path2)
    max_len = max(len(a), len(b))
    a = (a * ((max_len // len(a)) + 1))[:max_len]
    b = (b * ((max_len // len(b)) + 1))[:max_len]
    X, y = [], []
    for seq in a:
        X += [seq, reverse_complement(seq)]
        y += [1, 1]
    for seq in b:
        X += [seq, reverse_complement(seq)]
        y += [0, 0]
    return X, y

# --- Main run ---
if __name__ == "__main__":
    tokenizer = LongformerTokenizer.from_pretrained("allenai/longformer-base-4096")

    print("\n==== [Stage 1] Enhancer vs Non-Enhancer ====")
    X1, y1 = prepare_data("/kaggle/input/enhancer-detection-3/enhancer.cv.txt", "/kaggle/input/enhancer-detection-3/non.cv.txt")
    print("[Stage1] Label Distribution:", Counter(y1))
    train_kfold_with_plots(X1, y1, stage="Stage1", tokenizer=tokenizer)

    print("\n==== [Stage 2] Strong vs Weak Enhancer ====")
    X2, y2 = prepare_data("/kaggle/input/enhancer-detection-3/strong_742.txt", "/kaggle/input/enhancer-detection-3/weak_742.txt")
    print("[Stage2] Label Distribution:", Counter(y2))
    train_kfold_with_plots(X2, y2, stage="Stage2", tokenizer=tokenizer)
```
