```python
#only DNA BERT + simple classifier on CLS token 
import torch
import torch.nn as nn
import numpy as np
import time
import matplotlib.pyplot as plt
from transformers import AutoTokenizer, AutoModel
from torch.utils.data import Dataset, DataLoader
from torch.optim import AdamW
from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import accuracy_score, f1_score, confusion_matrix, roc_auc_score, matthews_corrcoef, recall_score, roc_curve, auc
from collections import Counter
import random, re
from torch.cuda.amp import autocast, GradScaler

# --- Reproducibility ---
def set_seed(seed=42):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)
    torch.backends.cudnn.deterministic = True

set_seed()

# --- Utilities ---
def clean_sequence(seq):
    return re.sub('[^ATCG]', '', seq.upper())

def load_fasta_clean(path):
    with open(path) as f:
        lines = f.readlines()
    return [clean_sequence(lines[i+1].strip()) for i in range(0, len(lines), 2)]

def reverse_complement(seq):
    return seq.translate(str.maketrans("ATCG", "TAGC"))[::-1]

def kmer_tokenize(seq, k=6):
    return " ".join([seq[i:i+k] for i in range(len(seq)-k+1)])

def augment_sequence(seq):
    bases = ['A','T','C','G']
    seq_list = list(seq)
    for i in range(len(seq)):
        if random.random() < 0.05:
            seq_list[i] = random.choice([b for b in bases if b != seq_list[i]])
    kmers = ["".join(seq_list[i:i+6]) for i in range(len(seq_list)-5)]
    random.shuffle(kmers)
    return " ".join(kmers)

# --- Dataset ---
class SimpleDataset(Dataset):
    def __init__(self, sequences, labels, tokenizer, max_len=512, augment=False):
        self.sequences = sequences
        self.labels = labels
        self.tokenizer = tokenizer
        self.max_len = max_len
        self.augment = augment

    def __len__(self):
        return len(self.sequences)

    def __getitem__(self, idx):
        seq = self.sequences[idx]
        label = self.labels[idx]
        if self.augment and random.random() < 0.5:
            seq = augment_sequence(seq)
        else:
            seq = kmer_tokenize(seq)
        encoding = self.tokenizer(seq, padding="max_length", truncation=True, max_length=self.max_len, return_tensors="pt")
        return {
            'input_ids': encoding['input_ids'].squeeze(0),
            'attention_mask': encoding['attention_mask'].squeeze(0),
            'label': torch.tensor(label, dtype=torch.float)
        }

# --- Model: DNA BERT only ---
class DNA_BERT_Only(nn.Module):
    def __init__(self, bert_name="zhihan1996/DNA_bert_6", dropout=0.3):
        super().__init__()
        self.bert = AutoModel.from_pretrained(bert_name)
        self.classifier = nn.Sequential(
            nn.Dropout(dropout),
            nn.Linear(self.bert.config.hidden_size, 1)
        )
    def forward(self, input_ids, attention_mask):
        outputs = self.bert(input_ids=input_ids, attention_mask=attention_mask)
        cls_token = outputs.pooler_output  # [batch_size, hidden_size]
        logits = self.classifier(cls_token)
        return logits.squeeze(1)

# --- Metrics ---
def compute_metrics(y_true, y_pred, y_prob):
    cm = confusion_matrix(y_true, y_pred)
    tn, fp, fn, tp = cm.ravel()
    return {
        'SN': recall_score(y_true, y_pred),
        'SP': tn / (tn + fp + 1e-8),
        'ACC': accuracy_score(y_true, y_pred),
        'F1': f1_score(y_true, y_pred),
        'MCC': matthews_corrcoef(y_true, y_pred),
        'AUC': roc_auc_score(y_true, y_prob)
    }

# --- Training Loop with loss & ROC plotting ---
def train_kfold_with_plots(X, y, stage, tokenizer, epochs=8, k=5):
    skf = StratifiedKFold(n_splits=k, shuffle=True, random_state=42)
    device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
    print(f"[{stage}] Using {torch.cuda.device_count()} GPU(s):", [torch.cuda.get_device_name(i) for i in range(torch.cuda.device_count())])
    scaler = GradScaler()

    all_fprs = []
    all_tprs = []
    all_aucs = []
    mean_fpr = np.linspace(0, 1, 100)

    for fold, (train_idx, val_idx) in enumerate(skf.split(X, y)):
        print(f"\n🔁 {stage} - Fold {fold+1}/{k}")
        train_ds = SimpleDataset([X[i] for i in train_idx], [y[i] for i in train_idx], tokenizer, augment=True)
        val_ds = SimpleDataset([X[i] for i in val_idx], [y[i] for i in val_idx], tokenizer)
        train_loader = DataLoader(train_ds, batch_size=32, shuffle=True)
        val_loader = DataLoader(val_ds, batch_size=64)

        model = DNA_BERT_Only()
        model = nn.DataParallel(model)
        model = model.to(device)

        optimizer = AdamW(model.parameters(), lr=1e-5)
        loss_fn = nn.BCEWithLogitsLoss()

        train_losses = []
        model.train()
        for epoch in range(epochs):
            epoch_loss = 0
            for batch in train_loader:
                optimizer.zero_grad()
                ids = batch['input_ids'].to(device)
                mask = batch['attention_mask'].to(device)
                label = batch['label'].to(device)
                with autocast():
                    logits = model(ids, mask)
                    loss = loss_fn(logits, label)
                scaler.scale(loss).backward()
                scaler.step(optimizer)
                scaler.update()
                epoch_loss += loss.item()
            avg_loss = epoch_loss / len(train_loader)
            train_losses.append(avg_loss)
            print(f"Epoch {epoch+1}/{epochs} Loss: {avg_loss:.4f}")

        # Plot Loss Curve for this fold
        plt.figure(figsize=(6,4))
        plt.plot(range(1, epochs+1), train_losses, marker='o')
        plt.title(f'{stage} Fold {fold+1} Training Loss')
        plt.xlabel('Epoch')
        plt.ylabel('Loss')
        plt.grid()
        plt.show()

        # Evaluation
        model.eval()
        y_true, y_prob, y_pred = [], [], []
        with torch.no_grad():
            for batch in val_loader:
                ids = batch['input_ids'].to(device)
                mask = batch['attention_mask'].to(device)
                label = batch['label'].to(device)
                with autocast():
                    probs = torch.sigmoid(model(ids, mask))
                y_true.extend(label.cpu().numpy())
                y_prob.extend(probs.cpu().numpy())
                y_pred.extend((probs > 0.5).int().cpu().numpy())

        metrics = compute_metrics(y_true, y_pred, y_prob)
        print(f"Fold {fold+1} Metrics:", metrics)

        # ROC curve for this fold
        fpr, tpr, _ = roc_curve(y_true, y_prob)
        roc_auc = auc(fpr, tpr)
        all_fprs.append(np.interp(mean_fpr, fpr, tpr))
        all_fprs[-1][0] = 0.0
        all_aucs.append(roc_auc)

        plt.figure(figsize=(6,6))
        plt.plot(fpr, tpr, label=f'ROC fold {fold+1} (AUC = {roc_auc:.3f})')
        plt.plot([0,1], [0,1], linestyle='--', color='gray')
        plt.xlabel('False Positive Rate')
        plt.ylabel('True Positive Rate')
        plt.title(f'{stage} Fold {fold+1} ROC Curve')
        plt.legend()
        plt.grid()
        plt.show()

    # Plot mean ROC curve
    mean_tpr = np.mean(all_fprs, axis=0)
    mean_tpr[-1] = 1.0
    mean_auc = auc(mean_fpr, mean_tpr)
    std_auc = np.std(all_aucs)

    plt.figure(figsize=(7,7))
    plt.plot(mean_fpr, mean_tpr, color='b',
             label=f'Mean ROC (AUC = {mean_auc:.3f} ± {std_auc:.3f})', lw=2)
    plt.plot([0,1], [0,1], linestyle='--', color='gray')
    plt.xlabel('False Positive Rate')
    plt.ylabel('True Positive Rate')
    plt.title(f'{stage} Mean ROC Curve')
    plt.legend(loc='lower right')
    plt.grid()
    plt.show()

    print(f"\n[{stage}] Mean AUC: {mean_auc:.4f} ± {std_auc:.4f}")

# --- Data Preparation (same as yours) ---
def prepare_data(path1, path2):
    a = load_fasta_clean(path1)
    b = load_fasta_clean(path2)
    max_len = max(len(a), len(b))
    a = (a * ((max_len // len(a)) + 1))[:max_len]
    b = (b * ((max_len // len(b)) + 1))[:max_len]
    X, y = [], []
    for seq in a:
        X += [seq, reverse_complement(seq)]
        y += [1, 1]
    for seq in b:
        X += [seq, reverse_complement(seq)]
        y += [0, 0]
    return X, y

# --- Main run ---
if __name__ == "__main__":
    tokenizer = AutoTokenizer.from_pretrained("zhihan1996/DNA_bert_6", do_lower_case=False, trust_remote_code=True)

    print("\n==== [Stage 1] Enhancer vs Non-Enhancer ====")
    X1, y1 = prepare_data("/kaggle/input/tahapu/enhancer.cv.txt", "/kaggle/input/tahapu/non.cv.txt")
    print("[Stage1] Label Distribution:", Counter(y1))
    train_kfold_with_plots(X1, y1, stage="Stage1", tokenizer=tokenizer)

    print("\n==== [Stage 2] Strong vs Weak Enhancer ====")
    X2, y2 = prepare_data("/kaggle/input/tahapu/strong_742.txt", "/kaggle/input/tahapu/weak_742.txt")
    print("[Stage2] Label Distribution:", Counter(y2))
    train_kfold_with_plots(X2, y2, stage="Stage2", tokenizer=tokenizer)
```

```python
import torch
import torch.nn as nn
import numpy as np
import time
import matplotlib.pyplot as plt
from transformers import AutoTokenizer, AutoModel
from torch.utils.data import Dataset, DataLoader
from torch.optim import AdamW
from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import accuracy_score, f1_score, confusion_matrix, roc_auc_score, matthews_corrcoef, recall_score, roc_curve, auc
from collections import Counter
from tqdm import tqdm
import random, re
from torch.cuda.amp import autocast, GradScaler

# --- Reproducibility ---
def set_seed(seed=42):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)
    torch.backends.cudnn.deterministic = True

set_seed()

# --- Utilities ---
def clean_sequence(seq):
    return re.sub('[^ATCG]', '', seq.upper())

def load_fasta_clean(path):
    with open(path) as f:
        lines = f.readlines()
    return [clean_sequence(lines[i+1].strip()) for i in range(0, len(lines), 2)]

def reverse_complement(seq):
    return seq.translate(str.maketrans("ATCG", "TAGC"))[::-1]

def kmer_tokenize(seq, k=6):
    return " ".join([seq[i:i+k] for i in range(len(seq)-k+1)])

def augment_sequence(seq):
    bases = ['A','T','C','G']
    seq_list = list(seq)
    for i in range(len(seq)):
        if random.random() < 0.05:
            seq_list[i] = random.choice([b for b in bases if b != seq_list[i]])
    kmers = ["".join(seq_list[i:i+6]) for i in range(len(seq_list)-5)]
    random.shuffle(kmers)
    return " ".join(kmers)

# --- Dataset ---
class SimpleDataset(Dataset):
    def __init__(self, sequences, labels, tokenizer, max_len=512, augment=False):
        self.sequences = sequences
        self.labels = labels
        self.tokenizer = tokenizer
        self.max_len = max_len
        self.augment = augment

    def __len__(self):
        return len(self.sequences)

    def __getitem__(self, idx):
        seq = self.sequences[idx]
        label = self.labels[idx]
        if self.augment and random.random() < 0.5:
            seq = augment_sequence(seq)
        else:
            seq = kmer_tokenize(seq)
        encoding = self.tokenizer(seq, padding="max_length", truncation=True, max_length=self.max_len, return_tensors="pt")
        return {
            'input_ids': encoding['input_ids'].squeeze(0),
            'attention_mask': encoding['attention_mask'].squeeze(0),
            'label': torch.tensor(label, dtype=torch.float)
        }

# --- Model: DNA BERT + CNN ---
class DNABERT_CNN(nn.Module):
    def __init__(self, bert_name="zhihan1996/DNA_bert_6", hidden_dim=128, dropout=0.3):
        super().__init__()
        self.bert = AutoModel.from_pretrained(bert_name)
        self.cnn = nn.Sequential(
            nn.Conv1d(self.bert.config.hidden_size, 64, kernel_size=5, padding=2),
            nn.ReLU(),
            nn.MaxPool1d(kernel_size=2)
        )
        self.classifier = nn.Sequential(
            nn.Linear(64, 128),
            nn.ReLU(),
            nn.Dropout(dropout),
            nn.Linear(128, 1)
        )

    def forward(self, input_ids, attention_mask):
        outputs = self.bert(input_ids=input_ids, attention_mask=attention_mask)
        hidden_states = outputs.last_hidden_state  # [batch_size, seq_len, hidden_size]
        
        # Transpose hidden states for CNN input: [batch_size, hidden_size, seq_len]
        cnn_input = hidden_states.permute(0, 2, 1)
        cnn_out = self.cnn(cnn_input)
        cnn_out = torch.max(cnn_out, dim=2)[0]  # Max pooling across the sequence dimension

        # Pass through classifier
        logits = self.classifier(cnn_out)
        return logits.squeeze(1)

# --- Metrics ---
def compute_metrics(y_true, y_pred, y_prob):
    cm = confusion_matrix(y_true, y_pred)
    tn, fp, fn, tp = cm.ravel()
    return {
        'SN': recall_score(y_true, y_pred),
        'SP': tn / (tn + fp + 1e-8),
        'ACC': accuracy_score(y_true, y_pred),
        'F1': f1_score(y_true, y_pred),
        'MCC': matthews_corrcoef(y_true, y_pred),
        'AUC': roc_auc_score(y_true, y_prob)
    }

# --- Training Loop ---
def train_kfold(X, y, stage, tokenizer, epochs=8, k=5):
    skf = StratifiedKFold(n_splits=k, shuffle=True, random_state=42)
    device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
    print(f"[{stage}] Using {torch.cuda.device_count()} GPU(s):", [torch.cuda.get_device_name(i) for i in range(torch.cuda.device_count())])
    scaler = GradScaler()

    for fold, (train_idx, val_idx) in enumerate(skf.split(X, y)):
        print(f"\n🔁 {stage} - Fold {fold+1}/{k}")
        train_ds = SimpleDataset([X[i] for i in train_idx], [y[i] for i in train_idx], tokenizer, augment=True)
        val_ds = SimpleDataset([X[i] for i in val_idx], [y[i] for i in val_idx], tokenizer)
        train_loader = DataLoader(train_ds, batch_size=32, shuffle=True)
        val_loader = DataLoader(val_ds, batch_size=64)

        model = DNABERT_CNN()
        model = nn.DataParallel(model)
        model = model.to(device)

        optimizer = AdamW(model.parameters(), lr=1e-5)
        loss_fn = nn.BCEWithLogitsLoss()

        start_time = time.time()
        for epoch in range(epochs):
            model.train()
            total_loss = 0
            for batch in train_loader:
                optimizer.zero_grad()
                ids = batch['input_ids'].to(device)
                mask = batch['attention_mask'].to(device)
                label = batch['label'].to(device)
                with autocast():
                    logits = model(ids, mask)
                    loss = loss_fn(logits, label)
                scaler.scale(loss).backward()
                scaler.step(optimizer)
                scaler.update()
                total_loss += loss.item()
            print(f"Epoch {epoch+1} Loss: {total_loss/len(train_loader):.4f}")
        print(f"Training Time (Fold {fold+1}): {time.time() - start_time:.2f} sec")

        model.eval()
        y_true, y_prob, y_pred = [], [], []
        with torch.no_grad():
            infer_start = time.time()
            for batch in val_loader:
                ids = batch['input_ids'].to(device)
                mask = batch['attention_mask'].to(device)
                label = batch['label'].to(device)
                with autocast():
                    probs = torch.sigmoid(model(ids, mask))
                y_true.extend(label.cpu().numpy())
                y_prob.extend(probs.cpu().numpy())
                y_pred.extend((probs > 0.5).int().cpu().numpy())
            infer_time = time.time() - infer_start
        print(f"Inference Time (Fold {fold+1}): {infer_time:.2f} sec")
        print(f"Fold {fold+1} Metrics:", compute_metrics(y_true, y_pred, y_prob))

# --- Run ---
def prepare_data(path1, path2):
    a = load_fasta_clean(path1)
    b = load_fasta_clean(path2)
    max_len = max(len(a), len(b))
    a = (a * ((max_len // len(a)) + 1))[:max_len]
    b = (b * ((max_len // len(b)) + 1))[:max_len]
    X, y = [], []
    for seq in a:
        X += [seq, reverse_complement(seq)]
        y += [1, 1]
    for seq in b:
        X += [seq, reverse_complement(seq)]
        y += [0, 0]
    return X, y

if __name__ == "__main__":
    tokenizer = AutoTokenizer.from_pretrained("zhihan1996/DNA_bert_6", do_lower_case=False, trust_remote_code=True)

    print("\n==== [Stage 1] Enhancer vs Non-Enhancer ====")
    X1, y1 = prepare_data("/kaggle/input/tahapu/enhancer.cv.txt", "/kaggle/input/tahapu/non.cv.txt")
    print("[Stage1] Label Distribution:", Counter(y1))
    train_kfold(X1, y1, stage="Stage1", tokenizer=tokenizer)

    print("\n==== [Stage 2] Strong vs Weak Enhancer ====")
    X2, y2 = prepare_data("/kaggle/input/tahapu/strong_742.txt", "/kaggle/input/tahapu/weak_742.txt")
    print("[Stage2] Label Distribution:", Counter(y2))
    train_kfold(X2, y2, stage="Stage2", tokenizer=tokenizer)
```

```python
import torch
import torch.nn as nn
import numpy as np
import time
import matplotlib.pyplot as plt
from transformers import AutoTokenizer, AutoModel
from torch.utils.data import Dataset, DataLoader
from torch.optim import AdamW
from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import accuracy_score, f1_score, confusion_matrix, roc_auc_score, matthews_corrcoef, recall_score, roc_curve, auc
from collections import Counter
from tqdm import tqdm
import random, re
from torch.cuda.amp import autocast, GradScaler

# --- Reproducibility ---
def set_seed(seed=42):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)
    torch.backends.cudnn.deterministic = True

set_seed()

# --- Utilities ---
def clean_sequence(seq):
    return re.sub('[^ATCG]', '', seq.upper())

def load_fasta_clean(path):
    with open(path) as f:
        lines = f.readlines()
    return [clean_sequence(lines[i+1].strip()) for i in range(0, len(lines), 2)]

def reverse_complement(seq):
    return seq.translate(str.maketrans("ATCG", "TAGC"))[::-1]

def kmer_tokenize(seq, k=6):
    return " ".join([seq[i:i+k] for i in range(len(seq)-k+1)])

def augment_sequence(seq):
    bases = ['A','T','C','G']
    seq_list = list(seq)
    for i in range(len(seq)):
        if random.random() < 0.05:
            seq_list[i] = random.choice([b for b in bases if b != seq_list[i]])
    kmers = ["".join(seq_list[i:i+6]) for i in range(len(seq_list)-5)]
    random.shuffle(kmers)
    return " ".join(kmers)

# --- Dataset ---
class SimpleDataset(Dataset):
    def __init__(self, sequences, labels, tokenizer, max_len=512, augment=False):
        self.sequences = sequences
        self.labels = labels
        self.tokenizer = tokenizer
        self.max_len = max_len
        self.augment = augment

    def __len__(self):
        return len(self.sequences)

    def __getitem__(self, idx):
        seq = self.sequences[idx]
        label = self.labels[idx]
        if self.augment and random.random() < 0.5:
            seq = augment_sequence(seq)
        else:
            seq = kmer_tokenize(seq)
        encoding = self.tokenizer(seq, padding="max_length", truncation=True, max_length=self.max_len, return_tensors="pt")
        return {
            'input_ids': encoding['input_ids'].squeeze(0),
            'attention_mask': encoding['attention_mask'].squeeze(0),
            'label': torch.tensor(label, dtype=torch.float)
        }

# --- Model: DNA BERT + BiLSTM ---
class DNABERT_BiLSTM(nn.Module):
    def __init__(self, bert_name="zhihan1996/DNA_bert_6", hidden_dim=128, dropout=0.3):
        super().__init__()
        self.bert = AutoModel.from_pretrained(bert_name)
        self.lstm = nn.LSTM(self.bert.config.hidden_size, hidden_dim, batch_first=True, bidirectional=True)
        self.classifier = nn.Sequential(
            nn.Linear(hidden_dim * 2, 128),
            nn.ReLU(),
            nn.Dropout(dropout),
            nn.Linear(128, 1)
        )

    def forward(self, input_ids, attention_mask):
        outputs = self.bert(input_ids=input_ids, attention_mask=attention_mask)
        hidden_states = outputs.last_hidden_state  # [batch_size, seq_len, hidden_size]
        
        lstm_out, _ = self.lstm(hidden_states)  # [batch_size, seq_len, hidden_dim*2]
        lstm_out = torch.max(lstm_out, dim=1)[0]  # Max pooling over the sequence dimension

        logits = self.classifier(lstm_out)
        return logits.squeeze(1)

# --- Metrics ---
def compute_metrics(y_true, y_pred, y_prob):
    cm = confusion_matrix(y_true, y_pred)
    tn, fp, fn, tp = cm.ravel()
    return {
        'SN': recall_score(y_true, y_pred),
        'SP': tn / (tn + fp + 1e-8),
        'ACC': accuracy_score(y_true, y_pred),
        'F1': f1_score(y_true, y_pred),
        'MCC': matthews_corrcoef(y_true, y_pred),
        'AUC': roc_auc_score(y_true, y_prob)
    }

# --- Training Loop ---
def train_kfold(X, y, stage, tokenizer, epochs=8, k=5):
    skf = StratifiedKFold(n_splits=k, shuffle=True, random_state=42)
    device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
    print(f"[{stage}] Using {torch.cuda.device_count()} GPU(s):", [torch.cuda.get_device_name(i) for i in range(torch.cuda.device_count())])
    scaler = GradScaler()

    for fold, (train_idx, val_idx) in enumerate(skf.split(X, y)):
        print(f"\n🔁 {stage} - Fold {fold+1}/{k}")
        train_ds = SimpleDataset([X[i] for i in train_idx], [y[i] for i in train_idx], tokenizer, augment=True)
        val_ds = SimpleDataset([X[i] for i in val_idx], [y[i] for i in val_idx], tokenizer)
        train_loader = DataLoader(train_ds, batch_size=32, shuffle=True)
        val_loader = DataLoader(val_ds, batch_size=64)

        model = DNABERT_BiLSTM()
        model = nn.DataParallel(model)
        model = model.to(device)

        optimizer = AdamW(model.parameters(), lr=1e-5)
        loss_fn = nn.BCEWithLogitsLoss()

        start_time = time.time()
        for epoch in range(epochs):
            model.train()
            total_loss = 0
            for batch in train_loader:
                optimizer.zero_grad()
                ids = batch['input_ids'].to(device)
                mask = batch['attention_mask'].to(device)
                label = batch['label'].to(device)
                with autocast():
                    logits = model(ids, mask)
                    loss = loss_fn(logits, label)
                scaler.scale(loss).backward()
                scaler.step(optimizer)
                scaler.update()
                total_loss += loss.item()
            print(f"Epoch {epoch+1} Loss: {total_loss/len(train_loader):.4f}")
        print(f"Training Time (Fold {fold+1}): {time.time() - start_time:.2f} sec")

        model.eval()
        y_true, y_prob, y_pred = [], [], []
        with torch.no_grad():
            infer_start = time.time()
            for batch in val_loader:
                ids = batch['input_ids'].to(device)
                mask = batch['attention_mask'].to(device)
                label = batch['label'].to(device)
                with autocast():
                    probs = torch.sigmoid(model(ids, mask))
                y_true.extend(label.cpu().numpy())
                y_prob.extend(probs.cpu().numpy())
                y_pred.extend((probs > 0.5).int().cpu().numpy())
            infer_time = time.time() - infer_start
        print(f"Inference Time (Fold {fold+1}): {infer_time:.2f} sec")
        print(f"Fold {fold+1} Metrics:", compute_metrics(y_true, y_pred, y_prob))

# --- Run ---
def prepare_data(path1, path2):
    a = load_fasta_clean(path1)
    b = load_fasta_clean(path2)
    max_len = max(len(a), len(b))
    a = (a * ((max_len // len(a)) + 1))[:max_len]
    b = (b * ((max_len // len(b)) + 1))[:max_len]
    X, y = [], []
    for seq in a:
        X += [seq, reverse_complement(seq)]
        y += [1, 1]
    for seq in b:
        X += [seq, reverse_complement(seq)]
        y += [0, 0]
    return X, y

if __name__ == "__main__":
    tokenizer = AutoTokenizer.from_pretrained("zhihan1996/DNA_bert_6", do_lower_case=False, trust_remote_code=True)

    print("\n==== [Stage 1] Enhancer vs Non-Enhancer ====")
    X1, y1 = prepare_data("/kaggle/input/tahapu/enhancer.cv.txt", "/kaggle/input/tahapu/non.cv.txt")
    print("[Stage1] Label Distribution:", Counter(y1))
    train_kfold(X1, y1, stage="Stage1", tokenizer=tokenizer)

    print("\n==== [Stage 2] Strong vs Weak Enhancer ====")
    X2, y2 = prepare_data("/kaggle/input/tahapu/strong_742.txt", "/kaggle/input/tahapu/weak_742.txt")
    print("[Stage2] Label Distribution:", Counter(y2))
    train_kfold(X2, y2, stage="Stage2", tokenizer=tokenizer)
```

```python
import torch
import torch.nn as nn
import numpy as np
import time
import matplotlib.pyplot as plt
from transformers import AutoTokenizer, AutoModel
from torch.utils.data import Dataset, DataLoader
from torch.optim import AdamW
from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import accuracy_score, f1_score, confusion_matrix, roc_auc_score, matthews_corrcoef, recall_score, roc_curve, auc
from collections import Counter
from tqdm import tqdm
import random, re
from torch.cuda.amp import autocast, GradScaler

# --- Reproducibility ---
def set_seed(seed=42):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)
    torch.backends.cudnn.deterministic = True

set_seed()

# --- Utilities ---
def clean_sequence(seq):
    return re.sub('[^ATCG]', '', seq.upper())

def load_fasta_clean(path):
    with open(path) as f:
        lines = f.readlines()
    return [clean_sequence(lines[i+1].strip()) for i in range(0, len(lines), 2)]

def reverse_complement(seq):
    return seq.translate(str.maketrans("ATCG", "TAGC"))[::-1]

def kmer_tokenize(seq, k=6):
    return " ".join([seq[i:i+k] for i in range(len(seq)-k+1)])

def augment_sequence(seq):
    bases = ['A','T','C','G']
    seq_list = list(seq)
    for i in range(len(seq)):
        if random.random() < 0.05:
            seq_list[i] = random.choice([b for b in bases if b != seq_list[i]])
    kmers = ["".join(seq_list[i:i+6]) for i in range(len(seq_list)-5)]
    random.shuffle(kmers)
    return " ".join(kmers)

# --- Dataset ---
class SimpleDataset(Dataset):
    def __init__(self, sequences, labels, tokenizer, max_len=512, augment=False):
        self.sequences = sequences
        self.labels = labels
        self.tokenizer = tokenizer
        self.max_len = max_len
        self.augment = augment

    def __len__(self):
        return len(self.sequences)

    def __getitem__(self, idx):
        seq = self.sequences[idx]
        label = self.labels[idx]
        if self.augment and random.random() < 0.5:
            seq = augment_sequence(seq)
        else:
            seq = kmer_tokenize(seq)
        encoding = self.tokenizer(seq, padding="max_length", truncation=True, max_length=self.max_len, return_tensors="pt")
        return {
            'input_ids': encoding['input_ids'].squeeze(0),
            'attention_mask': encoding['attention_mask'].squeeze(0),
            'label': torch.tensor(label, dtype=torch.float)
        }

# --- Model: DNA BERT + Transformer Encoder Layer(s) ---
class DNABERT_TransformerEncoder(nn.Module):
    def __init__(self, bert_name="zhihan1996/DNA_bert_6", hidden_dim=128, num_layers=2, dropout=0.3):
        super().__init__()
        self.bert = AutoModel.from_pretrained(bert_name)
        self.transformer_encoder_layer = nn.TransformerEncoderLayer(
            d_model=self.bert.config.hidden_size,
            nhead=8,  # Number of attention heads
            dim_feedforward=hidden_dim,
            dropout=dropout
        )
        self.transformer_encoder = nn.TransformerEncoder(
            self.transformer_encoder_layer, num_layers=num_layers
        )
        self.classifier = nn.Sequential(
            nn.Linear(self.bert.config.hidden_size, 128),
            nn.ReLU(),
            nn.Dropout(dropout),
            nn.Linear(128, 1)
        )

    def forward(self, input_ids, attention_mask):
        outputs = self.bert(input_ids=input_ids, attention_mask=attention_mask)
        hidden_states = outputs.last_hidden_state  # [batch_size, seq_len, hidden_size]
        
        # Pass through Transformer Encoder
        transformer_out = self.transformer_encoder(hidden_states)  # [batch_size, seq_len, hidden_size]

        # We can take the representation of the first token (CLS) or apply pooling
        cls_token = transformer_out[:, 0, :]  # [batch_size, hidden_size]
        
        logits = self.classifier(cls_token)
        return logits.squeeze(1)

# --- Metrics ---
def compute_metrics(y_true, y_pred, y_prob):
    cm = confusion_matrix(y_true, y_pred)
    tn, fp, fn, tp = cm.ravel()
    return {
        'SN': recall_score(y_true, y_pred),
        'SP': tn / (tn + fp + 1e-8),
        'ACC': accuracy_score(y_true, y_pred),
        'F1': f1_score(y_true, y_pred),
        'MCC': matthews_corrcoef(y_true, y_pred),
        'AUC': roc_auc_score(y_true, y_prob)
    }

# --- Training Loop ---
def train_kfold(X, y, stage, tokenizer, epochs=8, k=5):
    skf = StratifiedKFold(n_splits=k, shuffle=True, random_state=42)
    device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
    print(f"[{stage}] Using {torch.cuda.device_count()} GPU(s):", [torch.cuda.get_device_name(i) for i in range(torch.cuda.device_count())])
    scaler = GradScaler()

    for fold, (train_idx, val_idx) in enumerate(skf.split(X, y)):
        print(f"\n🔁 {stage} - Fold {fold+1}/{k}")
        train_ds = SimpleDataset([X[i] for i in train_idx], [y[i] for i in train_idx], tokenizer, augment=True)
        val_ds = SimpleDataset([X[i] for i in val_idx], [y[i] for i in val_idx], tokenizer)
        train_loader = DataLoader(train_ds, batch_size=32, shuffle=True)
        val_loader = DataLoader(val_ds, batch_size=64)

        model = DNABERT_TransformerEncoder()
        model = nn.DataParallel(model)
        model = model.to(device)

        optimizer = AdamW(model.parameters(), lr=1e-5)
        loss_fn = nn.BCEWithLogitsLoss()

        start_time = time.time()
        for epoch in range(epochs):
            model.train()
            total_loss = 0
            for batch in train_loader:
                optimizer.zero_grad()
                ids = batch['input_ids'].to(device)
                mask = batch['attention_mask'].to(device)
                label = batch['label'].to(device)
                with autocast():
                    logits = model(ids, mask)
                    loss = loss_fn(logits, label)
                scaler.scale(loss).backward()
                scaler.step(optimizer)
                scaler.update()
                total_loss += loss.item()
            print(f"Epoch {epoch+1} Loss: {total_loss/len(train_loader):.4f}")
        print(f"Training Time (Fold {fold+1}): {time.time() - start_time:.2f} sec")

        model.eval()
        y_true, y_prob, y_pred = [], [], []
        with torch.no_grad():
            infer_start = time.time()
            for batch in val_loader:
                ids = batch['input_ids'].to(device)
                mask = batch['attention_mask'].to(device)
                label = batch['label'].to(device)
                with autocast():
                    probs = torch.sigmoid(model(ids, mask))
                y_true.extend(label.cpu().numpy())
                y_prob.extend(probs.cpu().numpy())
                y_pred.extend((probs > 0.5).int().cpu().numpy())
            infer_time = time.time() - infer_start
        print(f"Inference Time (Fold {fold+1}): {infer_time:.2f} sec")
        print(f"Fold {fold+1} Metrics:", compute_metrics(y_true, y_pred, y_prob))

# --- Run ---
def prepare_data(path1, path2):
    a = load_fasta_clean(path1)
    b = load_fasta_clean(path2)
    max_len = max(len(a), len(b))
    a = (a * ((max_len // len(a)) + 1))[:max_len]
    b = (b * ((max_len // len(b)) + 1))[:max_len]
    X, y = [], []
    for seq in a:
        X += [seq, reverse_complement(seq)]
        y += [1, 1]
    for seq in b:
        X += [seq, reverse_complement(seq)]
        y += [0, 0]
    return X, y

if __name__ == "__main__":
    tokenizer = AutoTokenizer.from_pretrained("zhihan1996/DNA_bert_6", do_lower_case=False, trust_remote_code=True)

    print("\n==== [Stage 1] Enhancer vs Non-Enhancer ====")
    X1, y1 = prepare_data("/kaggle/input/tahapu/enhancer.cv.txt", "/kaggle/input/tahapu/non.cv.txt")
    print("[Stage1] Label Distribution:", Counter(y1))
    train_kfold(X1, y1, stage="Stage1", tokenizer=tokenizer)

    print("\n==== [Stage 2] Strong vs Weak Enhancer ====")
    X2, y2 = prepare_data("/kaggle/input/tahapu/strong_742.txt", "/kaggle/input/tahapu/weak_742.txt")
    print("[Stage2] Label Distribution:", Counter(y2))
    train_kfold(X2, y2, stage="Stage2", tokenizer=tokenizer)
```

**DNA BERT + CNN + BiLSTM + Attention**

```python
import torch
import torch.nn as nn
import numpy as np
import time
import matplotlib.pyplot as plt
from transformers import AutoTokenizer, AutoModel
from torch.utils.data import Dataset, DataLoader
from torch.optim import AdamW
from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import accuracy_score, f1_score, confusion_matrix, roc_auc_score, matthews_corrcoef, recall_score, roc_curve, auc
from collections import Counter
from tqdm import tqdm
import random, re, os, gc
from torch.cuda.amp import autocast, GradScaler

# --- Reproducibility ---
def set_seed(seed=42):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)
    torch.backends.cudnn.deterministic = True

set_seed()

# --- Utilities ---
def clean_sequence(seq):
    return re.sub('[^ATCG]', '', seq.upper())

def load_fasta_clean(path):
    with open(path) as f:
        lines = f.readlines()
    return [clean_sequence(lines[i+1].strip()) for i in range(0, len(lines), 2)]

def reverse_complement(seq):
    return seq.translate(str.maketrans("ATCG", "TAGC"))[::-1]

def kmer_tokenize(seq, k=6):
    return " ".join([seq[i:i+k] for i in range(len(seq)-k+1)])

def augment_sequence(seq):
    bases = ['A','T','C','G']
    seq_list = list(seq)
    for i in range(len(seq)):
        if random.random() < 0.05:
            seq_list[i] = random.choice([b for b in bases if b != seq_list[i]])
    kmers = ["".join(seq_list[i:i+6]) for i in range(len(seq_list)-5)]
    random.shuffle(kmers)
    return " ".join(kmers)

# --- Dataset ---
class SimpleDataset(Dataset):
    def __init__(self, sequences, labels, tokenizer, max_len=512, augment=False):
        self.sequences = sequences
        self.labels = labels
        self.tokenizer = tokenizer
        self.max_len = max_len
        self.augment = augment

    def __len__(self):
        return len(self.sequences)

    def __getitem__(self, idx):
        seq = self.sequences[idx]
        label = self.labels[idx]
        if self.augment and random.random() < 0.5:
            seq = augment_sequence(seq)
        else:
            seq = kmer_tokenize(seq)
        encoding = self.tokenizer(seq, padding="max_length", truncation=True, max_length=self.max_len, return_tensors="pt")
        return {
            'input_ids': encoding['input_ids'].squeeze(0),
            'attention_mask': encoding['attention_mask'].squeeze(0),
            'label': torch.tensor(label, dtype=torch.float)
        }

# --- Model: DNA BERT + CNN + BiLSTM + Attention ---
class EnhancedDNABERT(nn.Module):
    def __init__(self, bert_name="zhihan1996/DNA_bert_6", hidden_dim=128, dropout=0.3):
        super().__init__()
        self.bert = AutoModel.from_pretrained(bert_name)
        self.cnn = nn.Sequential(
            nn.Conv1d(self.bert.config.hidden_size, 64, kernel_size=5, padding=2),
            nn.ReLU(),
            nn.MaxPool1d(kernel_size=2)
        )
        self.lstm = nn.LSTM(self.bert.config.hidden_size, hidden_dim, batch_first=True, bidirectional=True)
        self.attn_fc = nn.Linear(hidden_dim * 2, 1)
        self.classifier = nn.Sequential(
            nn.Linear(hidden_dim*2 + 64 + self.bert.config.hidden_size, 128),
            nn.ReLU(),
            nn.Dropout(dropout),
            nn.Linear(128, 1)
        )

    def forward(self, input_ids, attention_mask):
        outputs = self.bert(input_ids=input_ids, attention_mask=attention_mask)
        cls_token = outputs.pooler_output
        hidden_states = outputs.last_hidden_state

        cnn_input = hidden_states.permute(0, 2, 1)
        cnn_out = self.cnn(cnn_input)
        cnn_out = torch.max(cnn_out, dim=2)[0]

        lstm_out, _ = self.lstm(hidden_states)
        attn_scores = torch.softmax(self.attn_fc(lstm_out).squeeze(-1), dim=1)
        attn_out = torch.sum(attn_scores.unsqueeze(-1) * lstm_out, dim=1)

        combined = torch.cat([cls_token, cnn_out, attn_out], dim=1)
        return self.classifier(combined).squeeze(1)

# --- Metrics ---
def compute_metrics(y_true, y_pred, y_prob):
    cm = confusion_matrix(y_true, y_pred)
    tn, fp, fn, tp = cm.ravel()
    return {
        'SN': recall_score(y_true, y_pred),
        'SP': tn / (tn + fp + 1e-8),
        'ACC': accuracy_score(y_true, y_pred),
        'F1': f1_score(y_true, y_pred),
        'MCC': matthews_corrcoef(y_true, y_pred),
        'AUC': roc_auc_score(y_true, y_prob)
    }

# --- Training Loop with Loss & ROC Curves ---
def train_kfold(X, y, stage, tokenizer, epochs=8, k=5):
    skf = StratifiedKFold(n_splits=k, shuffle=True, random_state=42)
    device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
    print(f"[{stage}] Using {torch.cuda.device_count()} GPU(s):", [torch.cuda.get_device_name(i) for i in range(torch.cuda.device_count())])
    scaler = GradScaler()

    all_fprs = []
    all_tprs = []
    all_aucs = []
    mean_fpr = np.linspace(0, 1, 100)

    for fold, (train_idx, val_idx) in enumerate(skf.split(X, y)):
        print(f"\n🔁 {stage} - Fold {fold+1}/{k}")
        train_ds = SimpleDataset([X[i] for i in train_idx], [y[i] for i in train_idx], tokenizer, augment=True)
        val_ds = SimpleDataset([X[i] for i in val_idx], [y[i] for i in val_idx], tokenizer)
        train_loader = DataLoader(train_ds, batch_size=32, shuffle=True)
        val_loader = DataLoader(val_ds, batch_size=64)

        model = EnhancedDNABERT()
        model = nn.DataParallel(model)
        model = model.to(device)

        optimizer = AdamW(model.parameters(), lr=1e-5)
        loss_fn = nn.BCEWithLogitsLoss()

        train_losses = []

        start_time = time.time()
        for epoch in range(epochs):
            epoch_loss = 0
            for batch in train_loader:
                optimizer.zero_grad()
                ids = batch['input_ids'].to(device)
                mask = batch['attention_mask'].to(device)
                label = batch['label'].to(device)
                with autocast():
                    logits = model(ids, mask)
                    loss = loss_fn(logits, label)
                scaler.scale(loss).backward()
                scaler.step(optimizer)
                scaler.update()
                epoch_loss += loss.item()
            avg_loss = epoch_loss / len(train_loader)
            train_losses.append(avg_loss)
            print(f"Epoch {epoch+1}/{epochs} Loss: {avg_loss:.4f}")

        # Plot Loss Curve for this fold
        plt.figure(figsize=(6,4))
        plt.plot(range(1, epochs+1), train_losses, marker='o')
        plt.title(f'{stage} Fold {fold+1} Training Loss')
        plt.xlabel('Epoch')
        plt.ylabel('Loss')
        plt.grid()
        plt.show()

        # Evaluation
        model.eval()
        y_true, y_prob, y_pred = [], [], []
        with torch.no_grad():
            for batch in val_loader:
                ids = batch['input_ids'].to(device)
                mask = batch['attention_mask'].to(device)
                label = batch['label'].to(device)
                with autocast():
                    probs = torch.sigmoid(model(ids, mask))
                y_true.extend(label.cpu().numpy())
                y_prob.extend(probs.cpu().numpy())
                y_pred.extend((probs > 0.5).int().cpu().numpy())

        metrics = compute_metrics(y_true, y_pred, y_prob)
        print(f"Fold {fold+1} Metrics:", metrics)

        # ROC curve for this fold
        fpr, tpr, _ = roc_curve(y_true, y_prob)
        roc_auc = auc(fpr, tpr)
        all_fprs.append(np.interp(mean_fpr, fpr, tpr))
        all_fprs[-1][0] = 0.0
        all_aucs.append(roc_auc)

        plt.figure(figsize=(6,6))
        plt.plot(fpr, tpr, label=f'ROC fold {fold+1} (AUC = {roc_auc:.3f})')
        plt.plot([0,1], [0,1], linestyle='--', color='gray')
        plt.xlabel('False Positive Rate')
        plt.ylabel('True Positive Rate')
        plt.title(f'{stage} Fold {fold+1} ROC Curve')
        plt.legend()
        plt.grid()
        plt.show()

    # Plot mean ROC curve
    mean_tpr = np.mean(all_fprs, axis=0)
    mean_tpr[-1] = 1.0
    mean_auc = auc(mean_fpr, mean_tpr)
    std_auc = np.std(all_aucs)

    plt.figure(figsize=(7,7))
    plt.plot(mean_fpr, mean_tpr, color='b',
             label=f'Mean ROC (AUC = {mean_auc:.3f} ± {std_auc:.3f})', lw=2)
    plt.plot([0,1], [0,1], linestyle='--', color='gray')
    plt.xlabel('False Positive Rate')
    plt.ylabel('True Positive Rate')
    plt.title(f'{stage} Mean ROC Curve')
    plt.legend(loc='lower right')
    plt.grid()
    plt.show()

    print(f"\n[{stage}] Mean AUC: {mean_auc:.4f} ± {std_auc:.4f}")

# --- Data Preparation ---
def prepare_data(path1, path2):
    a = load_fasta_clean(path1)
    b = load_fasta_clean(path2)
    max_len = max(len(a), len(b))
    a = (a * ((max_len // len(a)) + 1))[:max_len]
    b = (b * ((max_len // len(b)) + 1))[:max_len]
    X, y = [], []
    for seq in a:
        X += [seq, reverse_complement(seq)]
        y += [1, 1]
    for seq in b:
        X += [seq, reverse_complement(seq)]
        y += [0, 0]
    return X, y

# --- Main run ---
if __name__ == "__main__":
    tokenizer = AutoTokenizer.from_pretrained("zhihan1996/DNA_bert_6", do_lower_case=False, trust_remote_code=True)

    print("\n==== [Stage 1] Enhancer vs Non-Enhancer ====")
    X1, y1 = prepare_data("/kaggle/input/tahapu/enhancer.cv.txt", "/kaggle/input/tahapu/non.cv.txt")
    print("[Stage1] Label Distribution:", Counter(y1))
    train_kfold(X1, y1, stage="Stage1", tokenizer=tokenizer)

    print("\n==== [Stage 2] Strong vs Weak Enhancer ====")
    X2, y2 = prepare_data("/kaggle/input/tahapu/strong_742.txt", "/kaggle/input/tahapu/weak_742.txt")
    print("[Stage2] Label Distribution:", Counter(y2))
    train_kfold(X2, y2, stage="Stage2", tokenizer=tokenizer)
```
